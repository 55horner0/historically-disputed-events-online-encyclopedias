```python
import requests

def get_revision_history(title, lang="en", limit=50):
    url = f"https://{lang}.wikipedia.org/w/api.php"
    params = {
        "action": "query",
        "prop": "revisions",
        "titles": title,
        "rvlimit": limit,
        "rvprop": "ids|timestamp|user|comment|size|flags",
        "format": "json",
        "formatversion": 2,
    }
    headers = {"User-Agent": "my-research-tool/1.0 (contact: you@example.com)"}

    revisions = []
    while True:
        r = requests.get(url, params=params, headers=headers, timeout=15)

        # Debug: check status and content before parsing
        if r.status_code != 200:
            print("Status code:", r.status_code)
            print("Response text (first 500 chars):", r.text[:500])
            r.raise_for_status()

        try:
            data = r.json()
        except requests.exceptions.JSONDecodeError:
            print("Non-JSON response received. First 500 chars:")
            print(r.text[:500])
            raise

        page = data["query"]["pages"][0]
        revisions.extend(page.get("revisions", []))

        if "continue" in data:
            params.update(data["continue"])
        else:
            break
    return revisions

revs = get_revision_history("Climate change", limit=500)
for rv in revs[:20]:
    print(rv["timestamp"], rv["user"], rv["comment"])
```
