```python
import json
import numpy as np
import pandas as pd

from sklearn.metrics import precision_score, recall_score, fbeta_score, confusion_matrix, balanced_accuracy_score
from utilities.load_models import load_Qwen38B_embedding
from utilities.omission_3 import calculate_omission
```

```python
# manual
list_manual_section_pairs = [
    {
        "section_id_a": "es:Q9648:0",
        "section_id_b": "en:Q9648:0",
        "pair_id": "section_0",
        "alignment_score": "Manual"
    },
    {
        "section_id_a": "es:Q9648:1",
        "section_id_b": "en:Q9648:1",
        "pair_id": "section_1",
        "alignment_score": "Manual"
    },
    {
        "section_id_a": "es:Q9648:2",
        "section_id_b": "en:Q9648:2",
        "pair_id": "section_2",
        "alignment_score": "Manual"
    },
    {
        "section_id_a": "es:Q9648:3",
        "section_id_b": "en:Q9648:3",
        "pair_id": "section_3",
        "alignment_score": "Manual"
    },
    {
        "section_id_a": "es:Q9648:6",
        "section_id_b": "en:Q9648:7",
        "pair_id": "section_5",
        "alignment_score": "Manual"
    }
]
```

```python
gold = pd.read_csv("master_APs_equal_manual.csv")
print(f"Gold standard loaded: {len(gold)} pairs")
```

```python

with open("propositions_s0_en.json", "r", encoding="utf-8") as file:
    AP_en = json.load(file)

with open("propositions_s0_es.json", "r", encoding="utf-8") as file:
    AP_es = json.load(file)

gold = pd.read_csv("master_APs_equal_manual.csv")
model = load_Qwen38B_embedding()

def get_section_aps(AP, section_id):
    return [
        {"text": ap["text"]}
        for section in AP
        for ap in section
        if section_id in ap["proposition_id"]
    ]

all_scores = []
all_labels = []

for pair in list_manual_section_pairs:
    section_pair_id = pair["pair_id"]
    section_id_en = pair["section_id_b"]
    section_id_es = pair["section_id_a"]

    section_aps_en = get_section_aps(AP_en, section_id_en)
    section_aps_es = get_section_aps(AP_es, section_id_es)

    print(
        f"{section_id_en} (EN) vs {section_id_es} (ES) | "
        f"EN: {len(section_aps_en)} | ES: {len(section_aps_es)}"
    )

    omission_metrics = calculate_omission(
        section_aps_en,
        section_aps_es,
        threshold=0,
        model=model
    )

    matrix = np.asarray(omission_metrics["similarity_matrix"])

    golden_matrix = np.zeros(
        (len(section_aps_en), len(section_aps_es)),
        dtype=int
    )

    data_filtered = gold[gold["section_id"] == section_pair_id]

    for _, row in data_filtered.iterrows():
        en_index = int(row["en_index"])
        es_index = int(row["es_index"])
        golden_matrix[en_index, es_index] = 1

    all_scores.extend(matrix.flatten())
    all_labels.extend(golden_matrix.flatten())

all_scores = np.array(all_scores)
all_labels = np.array(all_labels)

print("\nDATASET")
print("Total pairs:", len(all_scores))
print("Equivalent:", np.sum(all_labels == 1))
print("Non-equivalent:", np.sum(all_labels == 0))

```

```python


threshold_results = []

for threshold in np.arange(0,1.001,0.001):
    predictions = (all_scores >= threshold).astype(int)

    precision = precision_score(all_labels,predictions,zero_division=0)
    recall = recall_score(all_labels,predictions,zero_division=0)
    f1 = fbeta_score(all_labels,predictions,beta=1,zero_division=0)
    balanced_acc = balanced_accuracy_score(all_labels,predictions)

    tn,fp,fn,tp = confusion_matrix(
        all_labels,
        predictions,
        labels=[0,1]
    ).ravel()

    fpr = fp/(fp+tn) if (fp+tn)>0 else 0

    threshold_results.append({
        "threshold":threshold,
        "precision":precision,
        "recall":recall,
        "F1":f1,
        "balanced_accuracy":balanced_acc,
        "FPR":fpr,
        "TP":tp,
        "FP":fp,
        "FN":fn,
        "TN":tn
    })

results_df = pd.DataFrame(threshold_results)

best_result = results_df.loc[results_df["F1"].idxmax()]

print("\nBEST THRESHOLD")
print(f"Threshold: {best_result['threshold']:.3f}")
print(f"Precision: {best_result['precision']:.3f}")
print(f"Recall: {best_result['recall']:.3f}")
print(f"F1: {best_result['F1']:.3f}")
print(f"Balanced accuracy: {best_result['balanced_accuracy']:.3f}")
print(f"FPR: {best_result['FPR']:.6f}")
print(f"TP: {int(best_result['TP'])}")
print(f"FP: {int(best_result['FP'])}")
print(f"FN: {int(best_result['FN'])}")
print(f"TN: {int(best_result['TN'])}")

print("\nTOP 20")

results_df.sort_values("F1",ascending=False).head(20)
```
